```python
import base64
import getpass

import cv2
import numpy as np
import pandas as pd

from google.colab import files
from IPython.display import Video, display

from inference_sdk import InferenceHTTPClient
from inference_sdk.webrtc import (
    VideoFileSource,
    StreamConfig,
    VideoMetadata
)

WORKSPACE = "nouras-workspace-rv8sa"

WORKFLOW = (
    "detect-masked-people-1-rfdetr-small-t1-"
    "vdetect-masked-people-1-rfdetr-small-t1-logic"
)

print("Libraries imported successfully!")
```

```python
print("Upload your video:")
uploaded = files.upload()

if not uploaded:
    raise ValueError("No video uploaded.")

video_path = next(iter(uploaded))

api_key = getpass.getpass("Roboflow API key: ")

if not api_key:
    raise ValueError("Roboflow API key required.")

client = InferenceHTTPClient.init(
    api_url="https://serverless.roboflow.com",
    api_key=api_key,
)

capture = cv2.VideoCapture(video_path)
source_fps = capture.get(cv2.CAP_PROP_FPS)
capture.release()

source_fps = source_fps if source_fps > 0 else 30.0

print("Video uploaded successfully!")
print("Connected to Roboflow!")
```

```python
source = VideoFileSource(
    video_path,
    realtime_processing=False
)

config = StreamConfig(
    stream_output=[],
    data_output=["output_image", "predictions"],
    realtime_processing=False,
)

session = client.webrtc.stream(
    source=source,
    workflow=WORKFLOW,
    workspace=WORKSPACE,
    image_input="image",
    config=config,
)

annotated_frames = []
prediction_rows = []
frame_errors = []

@session.on_data()
def collect_result(data: dict, metadata: VideoMetadata):
    if not isinstance(data, dict):
        return

    frame_id = int(metadata.frame_id)

    image_result = data.get("output_image")

    encoded = (
        image_result.get("value")
        if isinstance(image_result, dict)
        else image_result
    )

    if isinstance(encoded, str) and encoded:
        if encoded.startswith("data:image"):
            encoded = encoded.split(",", 1)[1]

        frame = cv2.imdecode(
            np.frombuffer(
                base64.b64decode(encoded),
                dtype=np.uint8
            ),
            cv2.IMREAD_COLOR,
        )

        if frame is not None:
            annotated_frames.append((frame_id, frame))

    raw_predictions = data.get("predictions")

    if isinstance(raw_predictions, dict):
        detections = raw_predictions.get("predictions", [])
    elif isinstance(raw_predictions, list):
        detections = raw_predictions
    else:
        detections = []

    counts = {
        "with_mask": 0,
        "without_mask": 0,
        "mask_weared_incorrect": 0,
    }

    for detection in detections:
        label = detection.get("class")
        if label in counts:
            counts[label] += 1

    prediction_rows.append({
        "frame_id": frame_id,
        "time_seconds": round(
            float(metadata.pts) * float(metadata.time_base),
            3
        ),
        **counts,
        "total_detections": len(detections),
    })

    if len(prediction_rows) % 50 == 0:
        print(
            f"Received {len(prediction_rows)} frame results"
        )


@session.on_error
def collect_error(errors, metadata: VideoMetadata):
    frame_errors.append(
        (metadata.frame_id, str(errors))
    )
    print(
        f"Frame {metadata.frame_id} failed: {errors}"
    )

print("Video processing configured!")
```

```python
# Cell 4 - Test Roboflow Model

import cv2
from IPython.display import display, Image

print("Starting model test...")

# Read the uploaded video
cap = cv2.VideoCapture(video_path)

success, frame = cap.read()
cap.release()

if not success:
    raise RuntimeError("Could not read the video.")

# Save the first frame
frame = cv2.resize(frame, (640, 360))
cv2.imwrite("test_frame.jpg", frame)

# Run the trained model
result = client.run_workflow(
    workspace_name=WORKSPACE,
    workflow_id=WORKFLOW,
    images={"image": "test_frame.jpg"},
    use_cache=False
)

# Display results
output = result[0] if isinstance(result, list) else result

print("Model test completed!")
print("Workflow outputs:", list(output.keys()))
print("Predictions:", str(output.get("predictions", "Not found"))[:1500])

display(Image(filename="test_frame.jpg"))
```

```python
import cv2
import numpy as np
import pandas as pd
import time

print("Starting video detection...")

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise RuntimeError("Could not open the video.")

fps = cap.get(cv2.CAP_PROP_FPS)
fps = fps if fps > 0 else 30

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Process 2 frames per second
interval = max(1, round(fps / 2))

output_fps = fps / interval

annotated_frames = []
prediction_rows = []
frame_errors = []

frame_id = 0

while True:
    success, frame = cap.read()

    if not success:
        break

    if frame_id % interval != 0:
        frame_id += 1
        continue

    # Resize while preserving aspect ratio
    height, width = frame.shape[:2]
    scale = min(640 / width, 640 / height)

    new_width = max(1, round(width * scale))
    new_height = max(1, round(height * scale))

    resized = cv2.resize(
        frame, (new_width, new_height)
    )

    cv2.imwrite("current_frame.jpg", resized)

    try:
        # Run Roboflow model
        result = client.run_workflow(
            workspace_name=WORKSPACE,
            workflow_id=WORKFLOW,
            images={"image": "current_frame.jpg"},
            use_cache=False
        )

        data = result[0] if isinstance(result, list) else result

        raw_predictions = data.get("predictions", [])

        if isinstance(raw_predictions, dict):
            detections = raw_predictions.get("predictions", [])
        elif isinstance(raw_predictions, list):
            detections = raw_predictions
        else:
            detections = []

        annotated = resized.copy()
        mask_count = 0

        for detection in detections:

            if detection.get("class") != "with_mask":
                continue

            x = float(detection["x"])
            y = float(detection["y"])
            w = float(detection["width"])
            h = float(detection["height"])

            x1 = int(x - w / 2)
            y1 = int(y - h / 2)
            x2 = int(x + w / 2)
            y2 = int(y + h / 2)

            cv2.rectangle(
                annotated,
                (x1, y1),
                (x2, y2),
                (0, 255, 0),
                2
            )

            confidence = detection.get("confidence", 0)

            cv2.putText(
                annotated,
                f"Mask {confidence:.2f}",
                (x1, max(20, y1 - 10)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 255, 0),
                2
            )

            mask_count += 1

        annotated_frames.append(annotated)

        prediction_rows.append({
            "frame_id": frame_id,
            "time_seconds": round(frame_id / fps, 2),
            "with_mask": mask_count
        })

        if len(annotated_frames) % 10 == 0:
            print(
                f"Processed {len(annotated_frames)} frames"
            )

        time.sleep(0.1)

    except Exception as e:
        frame_errors.append((frame_id, str(e)))
        print(f"Error in frame {frame_id}: {e}")

    frame_id += 1

cap.release()

print("Video detection completed!")
print("Processed frames:", len(annotated_frames))
print("Errors:", len(frame_errors))

if not annotated_frames:
    raise RuntimeError(
        "No frames processed. "
        f"Errors: {frame_errors[:3]}"
    )
```

```python
from google.colab import files
from IPython.display import Video, display
import pandas as pd
import cv2

height, width = annotated_frames[0].shape[:2]

output_video = "mask_detection_result.mp4"

writer = cv2.VideoWriter(
    output_video,
    cv2.VideoWriter_fourcc(*"mp4v"),
    output_fps,
    (width, height)
)

if not writer.isOpened():
    raise RuntimeError("Cannot create output video.")

for frame in annotated_frames:
    if frame.shape[:2] != (height, width):
        frame = cv2.resize(frame, (width, height))
    writer.write(frame)

writer.release()

df = pd.DataFrame(prediction_rows)
df.to_csv("mask_detection_results.csv", index=False)

print("Video saved successfully!")
print("CSV saved successfully!")

display(Video(output_video, embed=True, width=500))

files.download(output_video)
files.download("mask_detection_results.csv")
```
